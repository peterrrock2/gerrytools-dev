`ScatterPlot` shows the relationship between two quantities. This example compares the two
highest-BVAP district ranks in every plan of the demonstration ensemble.

```python
import json
from pathlib import Path

import numpy as np

from gerrytools.plotting import ScatterPlot

records = json.loads(Path("data/ga_congressional_ensemble.json").read_text(encoding="utf-8"))
ranked = np.sort(np.asarray([record["BVAP20"] for record in records]), axis=1)
second_highest = ranked[:, -2]
highest = ranked[:, -1]
```

```python
plot = ScatterPlot(legend=False)
plot.add_series(
    second_highest,
    highest,
    markerfacecolor="default_grey",
    markerfacealpha=0.45,
)
plot.show()
```

Every x value is paired with the y value in the same position. `xy_pairs=` accepts the same
data as a sequence of `(x, y)` tuples. Repeated `add_series()` calls add more point sets.

## Add one comparison plan

`add_point()` places a named observation over the existing series. The shared controls guide
shows how names become legend entries; the plot below keeps the legend disabled.

```python
plot.add_point(
    second_highest[0],
    highest[0],
    "Starting plan",
    markerfacecolor="alizarin",
    markersize=9,
)
plot.show()
```

Reference lines, bands, limits, labels, and legends are covered in
[shared statistical controls](options.ipynb).
