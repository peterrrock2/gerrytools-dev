`Histogram` shows how often values fall within each interval. This example counts the
districts at or above 50% Black voting-age population in each plan of the demonstration
ensemble.

```python
import json
from pathlib import Path

import numpy as np

from gerrytools.plotting import Histogram

records = json.loads(Path("data/ga_congressional_ensemble.json").read_text(encoding="utf-8"))
bvap = np.asarray([record["BVAP20"] for record in records])
majority_bvap_counts = (bvap >= 0.50).sum(axis=1)
```

These are integer outcomes, so `center_data_on_bin_edges()` and a bin width of one give each
possible count its own centered bar.

```python
plot = Histogram()
plot.add_dataset(majority_bvap_counts, edgecolor="black")
plot.center_data_on_bin_edges()
plot.set_bins_by_width(1)
plot.set_xticks([0, 1, 2])
plot.set_xlim(-0.5, 2.5)
plot.show()
```

## Add a comparison value

Reference lines and point markers distinguish a plan from the ensemble distribution without
turning it into a second histogram.

```python
starting_count = int((bvap[0] >= 0.50).sum())
plot.add_vertical_lines(starting_count, linecolor="cherryblossompink", linewidth=2)
plot.set_bins_by_width(1)
plot.set_xticks([0, 1, 2])
plot.set_xlim(-0.5, 2.5)
plot.show()
```

## Compare distributions

Repeated `add_dataset()` calls can overlay, group, stack, or outline distributions according
to `histtype`. Use `as_density()` when distributions with different sample sizes should be
compared by proportion instead of raw count.

```python
higher_threshold_counts = (bvap >= 0.55).sum(axis=1)

comparison = Histogram()
comparison.add_dataset(majority_bvap_counts, facecolor="default_grey")
comparison.add_dataset(
    higher_threshold_counts,
    facecolor="citizen_blue",
    facealpha=0.65,
)
comparison.center_data_on_bin_edges()
comparison.set_bins_by_width(1)
comparison.set_bins_by_width(1)
comparison.set_xticks([0, 1, 2])
comparison.set_xlim(-0.5, 2.5)
comparison.show()
```

`add_dataset()` accepts an iterable, NumPy array, Series, or one DataFrame column. `weights=`
plots an aggregated value-count table without expanding it into repeated observations. Shared
axes, labels, legends, and output controls are covered in
[shared statistical controls](options.ipynb).
