`PaintballPlot` shows many observations in statewide vote-share and seat-share space. The
points may represent plans, elections, or modeled scenarios, as long as both sequences refer
to the same observations in the same order.

```python
import numpy as np

from gerrytools.plotting import PaintballPlot

rng = np.random.default_rng(12)
vote_shares = np.clip(rng.normal(0.50, 0.035, size=250), 0, 1)
seat_counts = np.clip(
    np.rint(7 + 24 * (vote_shares - 0.50) + rng.normal(0, 1.2, size=250)),
    0,
    14,
)

plot = PaintballPlot(legend=False)
plot.add_seats_votes_data(vote_shares, seat_counts, total_seats=14)
plot.set_marker_options(color="denim", alpha=0.55, size=5)
plot.show()
```

Pass seat shares directly, or pass seat counts with `total_seats=` as above. The plot
complements both supplied shares before drawing them: an input point `(0.48, 5 / 14)` appears
at `(0.52, 9 / 14)`. This convention lets Democratic input data be displayed as Republican
vote and seat shares.

## Reference lines

The proportionality and efficiency-gap helpers add common vote-seat references. Arbitrary
lines through `(0.5, 0.5)` can be added with `add_lines_with_slope()`.

```python
plot.add_proportionality_line(linecolor="default_grey", linestyle="--")
plot.show()
```

## Show the envelope

`display_hull(True)` replaces the points with their horizontal envelope. At each observed
seat-share level, it keeps the minimum and maximum vote-share coordinates and connects those
bounds.

```python
plot.display_hull(True)
plot.set_hull_options(color="lightblue", alpha=0.3, edgecolor="denim")
plot.show()
```
