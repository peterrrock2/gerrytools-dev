`SeaLevelPlot` connects one plan's values across an ordered set of categories. Repeating that
operation for many plans creates the characteristic sea of lines.

```python
from gerrytools.plotting import SeaLevelPlot

starting_plan = {
    "President 2016": 7 / 14,
    "Governor 2018": 6 / 14,
    "President 2020": 7 / 14,
    "U.S. Senate 2020": 8 / 14,
}

plot = SeaLevelPlot()
plot.add_dataset(starting_plan, linecolor="alizarin")
plot.show()
```

A mapping keeps category labels and order with the values. Lists, Series objects, and one
DataFrame row are also accepted.

## Add an ensemble

The following synthetic data give 80 plans a seat share for the same four elections. Adding
each plan as a low-alpha dataset reveals where the ensemble lines concentrate.

```python
import numpy as np

rng = np.random.default_rng(7)
expected_seats = np.array([7, 6, 8, 7])
ensemble_seats = np.clip(
    expected_seats + rng.integers(-2, 3, size=(80, 4)),
    0,
    14,
)

sea = SeaLevelPlot(jitter_rng_seed=7)
for seats in ensemble_seats:
    sea.add_dataset(
        (seats / 14).tolist(),
        category_labels=list(starting_plan),
        linecolor="denim",
        linealpha=0.08,
        markersize=0,
    )
sea.add_dataset(starting_plan, linecolor="alizarin", linewidth=2)
sea.show()
```

Seat shares occur in multiples of `1 / 14`, so many lines coincide. A small vertical jitter
can separate them without changing the underlying values. The seed makes the displacement
stable when the builder renders again.

```python
sea.set_vertical_jitter(0.01)
sea.show()
```
