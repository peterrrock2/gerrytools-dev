```python
import geopandas as gpd
from gerrytools.plotting.geoplot import GeoPlot, LabelFontOptions, LabelBoxOptions
```

```python
gdf = gpd.read_file("PA_VTD")
```

```python
gp = GeoPlot(gdf)
gp.add_districting_plan_layer(
    plancolumn="538GOP",
    colormap="districtr",
    facealpha=0.8,
    show_labels=True,
    exclude_labels=[1,2]
)

gp.show()
```

```python
gp = GeoPlot(gdf)
gp.add_choropleth_layer(
    datacolumn="TOTPOP",
    colormap="Greys"
)
gp.add_districting_plan_layer(
    plancolumn="538GOP",
    colormap="districtr",
    facealpha=0.8,
    show_labels=True
)
gp.add_outline_layer(
    geosource=gdf.loc[gdf["COUNTYFP10"].isin(["027", "029"])],
    dissolve_column="COUNTYFP10",
    show_labels=True,
    labelfont_options=LabelFontOptions(
        fontcolor="black",
        fontsize=4,
        fontweight="roman", 
        outlinecolor="grey",
        outlinewidth=0.2
    ),
    labelbox_options=LabelBoxOptions(
        facecolor="ivory",
        facealpha=0.5,
    )
)
gp.add_highlight_layer(
    geosource=gdf.loc[gdf["COUNTYFP10"].isin(["003"])],
)

gp.show()
```

```python
gp = GeoPlot(gdf)
gp.add_districting_plan_layer(
    plancolumn="538GOP",
    colormap="districtr",
    facealpha=0.8,
    show_labels=True
)
gp.add_outline_layer(
    geosource=gdf.loc[gdf["COUNTYFP10"].isin(["027", "029"])],
    dissolve_column="COUNTYFP10",
    show_labels=True,
)
gp.add_highlight_layer(
    geosource=gdf.loc[gdf["COUNTYFP10"].isin(["003"])],
    dissolve_column="COUNTYFP10",
    show_labels=True,
)
gp.focus_axes(
    geosource=gdf.loc[gdf["COUNTYFP10"].isin(["003"])],
)
gp.show()
```

```python
gp = GeoPlot(gdf, show_axis=True)

gp.add_choropleth_layer(
    datacolumn="TOTPOP",
    colormap="Greens",
    bins=10
)
gp.set_show_colorbars(True)
gp.show()
```

```python
gp = GeoPlot(gdf)

gp.add_choropleth_layer(
    datacolumn="TOTPOP",
    colormap="Greens",
    bins=[0,1000,2000,3000,4000,5000,6000,7000,8000,9000]
)
gp.set_colorbar_options(
    tick_fontsize=7,
    tick_pad=1,
    label_fontsize=8,
    outer_pad=0.02,
    inner_pad=0.04,
    width=0.018,
)
gp.set_show_colorbars(True)
gp.show()
```
