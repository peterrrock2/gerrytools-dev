```python
import geopandas as gpd
import pandas as pd
import numpy as np
from gerrytools.plotting.scatterplot import ScatterPlot
from gerrytools.plotting.gerryplot import GerryPlotBase
```

```python
def make_random_points(n_points = 1000):
    return np.random.random(size=n_points), np.random.random(size=n_points)
```

```python
plot = ScatterPlot()
x,y = make_random_points()
plot.add_scatter(x,y)
plot.show()
```
